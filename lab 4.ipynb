```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define Activation Functions and Derivatives
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh(x):
    return np.tanh(x)

def tanh_derivative(x):
    return 1.0 - np.tanh(x)**2

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1.0, 0.0)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# 2. Quantitative Evaluation Across Range [-5 to +5]
test_vals = np.array([-5.0, -2.5, 0.0, 2.5, 5.0])

print("=== Activation Function Evaluation across Range [-5 to +5] ===")
print("Input (x):     ", " ".join(f"{v:8.4f}" for v in test_vals))
print("Sigmoid(x):    ", " ".join(f"{sigmoid(v):8.4f}" for v in test_vals))
print("Sigmoid'(x):   ", " ".join(f"{sigmoid_derivative(v):8.4f}" for v in test_vals))
print("Tanh(x):       ", " ".join(f"{tanh(v):8.4f}" for v in test_vals))
print("Tanh'(x):      ", " ".join(f"{tanh_derivative(v):8.4f}" for v in test_vals))
print("ReLU(x):       ", " ".join(f"{relu(v):8.4f}" for v in test_vals))
print("ReLU'(x):      ", " ".join(f"{relu_derivative(v):8.4f}" for v in test_vals))
print("LeakyReLU(x):  ", " ".join(f"{leaky_relu(v):8.4f}" for v in test_vals))

# 3. Vanishing Gradient Simulation across 10 Deep Layers
n_layers = 10
grad_sigmoid, grad_tanh, grad_relu = 1.0, 1.0, 1.0
x_sat = 2.0  # saturated region

for _ in range(n_layers):
    grad_sigmoid *= sigmoid_derivative(x_sat)
    grad_tanh *= tanh_derivative(x_sat)
    grad_relu *= relu_derivative(x_sat)

print(f"\n=== Vanishing Gradient Simulation ({n_layers} Deep Layers at x=2.0) ===")
print("Initial Output Gradient Injected: 1.000000")
print(f"Gradient reaching Layer 1 (Sigmoid) : {grad_sigmoid:.12f}")
print(f"Gradient reaching Layer 1 (Tanh)    : {grad_tanh:.12f}")
print(f"Gradient reaching Layer 1 (ReLU)    : {grad_relu:.12f}")
```
