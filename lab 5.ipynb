```python
import random
import time

# 1. Environment Class for 2x2 Grid Toy Problem
class Environment:
    def __init__(self):
        # Initialize 2x2 grid rooms with random Clean/Dirty state
        self.rooms = {
            "A": random.choice(["Clean", "Dirty"]),
            "B": random.choice(["Clean", "Dirty"]),
            "C": random.choice(["Clean", "Dirty"]),
            "D": random.choice(["Clean", "Dirty"])
        }
        self.agent_location = random.choice(list(self.rooms.keys()))
        self.score = 0

    def display(self):
        print("Environment State:")
        print(f"  A: {self.rooms['A']} | B: {self.rooms['B']}")
        print(f"  C: {self.rooms['C']} | D: {self.rooms['D']}")
        print(f"  Agent is at: {self.agent_location}")
        print(f"  Current Score: {self.score}")
        print("-" * 50)


# 2. Reflex Vacuum Cleaner Agent Class
class ReflexVacuumAgent:
    def __init__(self, env):
        self.env = env

    def perceive_and_act(self):
        location = self.env.agent_location
        status = self.env.rooms[location]

        # Rule 1: Clean if current room is dirty (+10 points)
        if status == "Dirty":
            print(f"Cleaning room {location}...")
            self.env.rooms[location] = "Clean"
            self.env.score += 10

        # Rule 2: Move if clean (-1 point movement penalty)
        else:
            adjacent_rooms = list(self.env.rooms.keys())
            next_room = random.choice([r for r in adjacent_rooms if r != location])
            print(f"Moving from {location} -> {next_room}")
            self.env.agent_location = next_room
            self.env.score -= 1


# 3. Execute Agent in Environment
random.seed(42)
env = Environment()
agent = ReflexVacuumAgent(env)

print("=== Starting Reflex Vacuum Agent Simulation ===")
env.display()

for step in range(1, 6):
    print(f"\nStep {step}:")
    agent.perceive_and_act()
    env.display()

print("All accessible rooms cleaned successfully!")
```
